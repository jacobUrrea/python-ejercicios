Las clases permiten representar entidades dentro del código.
En este cuaderno se definiran clases sencillas, añadir métodos, utilizar constructores, aplicar herencia y emplear métodos de clase y estáticos.


Caracteristicas:
- Clase: Plano o plantilla para crear objetos
- Objeto o Instancia: Ejemplar concreto de una clase
- Atributo: Variable asociada a un objeto o clase
- Método: Función asociada a un objeto o clase
- Constructor: Método especial que inicializa objetos

Ventajas de usar clases:
- Agrupa datos organizadamente
- La herencia permite extender su funcionalidad y reutilizar codigo
- Abstracción: Oculta detalles de implementación

Casos de uso comunes:
- Modelar entidades del mundo real (usuarios, productos, vehículos)
- Crear estructuras de datos complejas
- Implementar patrones de diseño
- Desarrollar frameworks y librerías

```python
#Sintaxis Basica
class NombreClase:
    def __init__(self, param1, param2):  # Constructor
        self.param1 = param1  # Atributo de instancia
        self.param2 = param2
    
    def metodo(self):  # Metodo
        return self.param1
```

self: Referencia al objeto actual, como "this" en otros lenguajes

__init__: Sirve como constructor o inicializador, no es un constructor estricto

```python
class Album:
    #Clase para un album musical
    def __init__(self, titulo, artista, año):
        self.titulo = titulo
        self.artista = artista
        self.año = año

# Creacion de instancia
a = Album("Thriller", "Michael Jackson", 1982)
print(a.titulo, "-", a.artista, "(", a.año, ")")
```

Se agrego un metodo que devuelve una cadena con información del álbum.

```python
class Album:
    def __init__(self, titulo, artista, año, ventas):
        self.titulo = titulo
        self.artista = artista
        self.año = año
        self.ventas = ventas
    def info(self):
        return f"{self.titulo} by {self.artista} ({self.año}) - {self.ventas}M de ventas"

    def __str__(self):
        return self.info()

# Prueba
t = Album('Back in Black', 'AC/DC', 1980, 50)
print(t)  # usa __str__
```
