En este cuaderno se desarrollan ejercicios prácticos sobre el uso de cadenas en Python, abordando operaciones básicas, métodos y manipulación de texto.

Un string en Python es una secuencia de caracteres, ya sea letras, números o símbolos; que se define entre comillas simples (' ') o dobles (" "). Representa un tipo de dato fundamental a la hora de representar un texto.
Presenta las siguientes caracteristicas:
- Se crea con comillas simples o dobles
- Se puede usar triple comilla (''' ''' o """ """) para strings de varias líneas
- Inmutabilidad: No se puede modificar después de creado, solo se puede reemplazar o eliminar completamente
- Indexación: Se accede a caracteres individuales con corchetes [ ], empieza desde 0
- Es posible obtener la longitud mediante len(string).
- Búsqueda: Se verifica contenido con in o not in.
- Se puede separar en partes con split().

¿Cual es el valor de la variable <code>A</code> después de ejecutar el siguiente código?

```python
A = "1"
print(A)
#La variable "A" almacena una cadena de texto, no un numero, por lo que su valor termina siendo el string "1"
```

¿Cual es el valor de la variable <code>B</code> después de ejecutar el siguiente código?

```python
B = "2"
print(B)
#Es igual que el ejemplo anterior, la variable "B" contiene la cadena "2"
```

¿Cual es el valor de la variable <code>C</code> después de ejecutar el siguiente código?

```python
C = A + B
print(C)
#¿Por que "A + B" devuelve "12" y no 3?  
#Porque A y B son strings, para una operacion aritmetica se necesita convertirlo a entero
```

En la variable <code>D</code> haz uso del slicing para imprimir solo los primeros tres elementos:

```python
D = "ABCDEFG"
print(D[0:3])
#El slicing [0:3] extrae los caracteres desde el indice 0 hasta el 2 
```

Usa un valor de stride de 2 para imprimir cada segundo elemento de la cadena <code>E</code>:

```python
E = "clocrkr1e1c1t"
print(E[::2])
#Cuando se usa el stride de 2, se seleccionara cada segundo caracter de la cadena
```

Imprime una barra invertida:

```python
print("\\")
#La barra invertida al ser un caracter especial, se tendra que escribir dos veces para imprimirla correctamente
```

Convierte el mayúsculas el contenido de la variable <code>F</code>:

```python
F = "You are wrong"
print(F.upper())
#Upper es un metodo que convierte todos los caracteres de la cadena a mayusculas
#Tambien existe Lower() usado para convertir los caracteres a minuscula
```

En la variable <code>G</code>, encuentra el primer índice de la sub-cadena <code>snow</code>:

```python
G = "Mary had a little lamb Little lamb, little lamb Mary had a little lamb \
Its fleece was white as snow And everywhere that Mary went Mary went, Mary went \
Everywhere that Mary went The lamb was sure to go"

print(G.find("snow"))
#Find es un metodo cuya funcion es devolver el indice de la primera aparicion de alguna sub-cadena, en este caso "snow"
```

En la variable <code>G</code>, reemplaza la sub-cadena <code>Mary</code> con <code>Bob</code>:

```python
print(G.replace("Mary", "Bob"))
#Replace es un metodo usado para sustituir todas las apariciones de "Mary" por "Bob" dentro de la cadena.
# En general toma la primera cadena como el valor a buscar y la segunda como el valor que la reemplaza.
```
