Caracteristicas: 
- Mismo output para mismos inputs
- Ejemplo: f(x) = x * x
- Inmutabilidad: Los datos no se modifican, se crean nuevos

Las funciones como objetos de primera clase:
- Pueden pasarse como argumentos
- Devolverse como valores
- Almacenarse en estructuras de datos

Cree una función que divida la primera entrada por la segunda entrada:

```python
#Funcion que recibe dos valores, a es el numerador y b es el denominador
#Se usa el / para que se realize la division
def dividir(a, b):
    return a / b
```

```python
resultado = dividir(20, 5)
resultado
```

Utilice la función <code> con </code> para la siguiente pregunta.

```python
def con(a, b):
    return(a + b)
```

¿Se puede usar la función <code> con </code> que definimos antes para agregar números enteros o cadenas?

```python
con(10, 5)
#Si los valores son numericos, Python lo interpreta como una suma matematica
```

```python
con("Hola ", "Mundo")
#Si los valores son cadenas de texto, Python lo interpreta como una concatenacion
```

```python
con("Hola ", 5)
#En este caso los valores son de diferente tipo, por lo que ocurrira un error al ejecutar el codigo
```

¿Se puede usar la función <code> con </code> que definimos antes para concentrar una lista o tupla?

```python
con([1, 2, 3], ["a", "b"])
#Python concatena listas cuando se usa el operador
```

```python
con((1, 2), ("a", "b"))
#Lo mismo ocurre con las tuplas
```

```python
con([1, 2], ("a", "b"))
#En Python no se permite sumar tipos distintos, en este caso una lista + tupla.
```

Las funciones en Python ofrecen la posibilidad de favorecer la reutilización de código de forma práctica y lógica.
El ejemplo de la función ilustra cómo es posible aprovechar el manejo de datos, permitiendo así que una misma función trabaje con diversos tipos de datos, como números, cadenas, listas o tuplas.
Sin embargo, esta flexibilidad está condicionada a que los tipos de datos utilizados sean compatibles entre sí para evitar errores en la ejecución.
