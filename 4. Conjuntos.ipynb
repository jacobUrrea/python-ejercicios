En este cuaderno se estudia el uso de conjuntos en Python, una estructura de datos cuya función es almacenar elementos únicos.
Además, se revisarán operaciones como eliminación de duplicados, suma de elementos,
unión de conjuntos y verificación de subconjuntos, resaltando su utilidad para el manejo eficiente de datos


Los conjuntos en Python son estructuras diseñadas para guardar colecciones de elementos únicos, sin un orden determinado. Estos son útiles en gran manera cuando se necesita evitar duplicados, realizar operaciones de tipo matemático entre conjuntos o verificar relaciones como la unión, pertenencia y intersección.


Transformar la lista <code>['rap','house','electronic music', 'rap']</code> en un conjunto:

```python
lista = ['rap', 'house', 'electronic music', 'rap']
conjunto = set(lista)
conjunto
#Esto es bastante usado al momento de limpiar datos, antes de su analisis
#Python eliminara automaticamente los elementos duplicados
```

Considera que la lista <code>A = [1, 2, 2, 1]</code> y el conjunto <code>B = set([1, 2, 2, 1])</code> suman <code>sum(A) = sum(B)</code>

```python
A = [1, 2, 2, 1] #Definicion de la lista y del conjunto
B = set([1, 2, 2, 1])
sumaLista = sum(A) #Incluira valores repetidos
sumaConjunto = sum(B) #Solo sumara valores unicos 
#por lo que aunque los valores sean los mismos, el resultaod de la suma de los conjuntos eliminara duplicados
sumaLista, sumaConjunto
```

Crea un nuevo conjunto <code>album_set3</code> que sea la unión de <code>album_set1</code> y <code>album_set2</code>:

```python
#Se definen los dos conjuntos
album_set1 = set(["Thriller", 'AC/DC', 'Back in Black']) 
album_set2 = set([ "AC/DC", "Back in Black", "The Dark Side of the Moon"])
```

```python
# Se crea un nuevo conjunto con la union de los anteriormente creados
album_set3 = album_set1.union(album_set2)
album_set3
#Album_set3 combina conjuntos sin repetir los elementos
```

Averigua si <code>album_set1</code> es un subconjunto de <code>album_set3</code>:

```python
# Verificion de que si album_set1 es subconjunto de album_set3
subconjunto = album_set1.issubset(album_set3)
subconjunto
#Devolvera True si todos los elementos de un conjunto estan contenidos en el otro
```
