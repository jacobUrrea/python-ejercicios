#### MENOR DE DOS PARES: Escriba una función que devuelva el menor de dos números dados *si* ambos números son pares, pero devuelva el mayor si uno o ambos números son impares
    lesser_of_two_evens(2,4) --> 2
    lesser_of_two_evens(2,5) --> 5

```python
def lesser_of_two_evens(a,b):
    # Si ambos numeros son pares, se retorna el menor
    if a % 2 == 0 and b % 2 == 0:
        if a < b:
            return a
        else:
            return b
    # Si uno o ambos son impares, se retorna el mayor
    else:
        if a > b:
            return a
        else:
            return b
    #Se utiliza el operador "%" para determinar la paridad de los numeros y se aprovechan
```

```python
# Revisión
lesser_of_two_evens(2,4)
```

```python
# Revisión
lesser_of_two_evens(2,5)
```

#### GALLETAS DE ANIMALES: Escribir una función que toma una cadena de dos palabras y devuelve True si ambas palabras comienzan con la misma letra
    animal_crackers('Ciudad Calor') --> True
    animal_crackers('Lugar Frio') --> False

```python
def animal_crackers(text):
  # Separa la cadena en palabras
  palabras = text.split()
  # Se compara la primera letra de cada palabra
  return palabras[0][0].lower() == palabras[1][0].lower()
#Se usa lower() para evitar errores entre mayusculas y minusculas
```

```python
# Revisión
animal_crackers('Ciudad Calor')
```

```python
# Revisión
animal_crackers('Lugar Frio')
```

#### HACE VEINTE: dados dos enteros, devuelve True si la suma de los enteros es 20 o si uno de los enteros es 20. De lo contrario, devuelve False

    makes_twenty(20,10) --> True
    makes_twenty(12,8) --> True
    makes_twenty(2,3) --> False

```python
def makes_twenty(n1, n2):
    # Retorna True si uno de los enteros es 20 o si la suma es 20
    return n1 == 20 or n2 == 20 or (n1 + n2) == 20
```

```python
# Revisión
makes_twenty(20,10)
```

```python
# Revisión
makes_twenty(2,3)
```

```python
makes_twenty(12,8)
```

#### REVERSA: Dada una oración, retornar la misma oración pero en reversa

    master_yoda('I am home') --> 'home am I'
    master_yoda('We are ready') --> 'ready are We'
    
Note: Revisar método .join()

    >>> "--".join(['a','b','c'])
    >>> 'a--b--c'

Resultado de .join():

    >>> " ".join(['Hello','world'])
    >>> "Hello world"

```python
def master_yoda(text):
    palabras = text.split()
    nuevapalabra = palabras[::-1]

    return " ".join(nuevapalabra)

#El uso combinado de split(), slicing y join() permite resolver el problema de una manera sencilla
```

```python
# Revisión
master_yoda('I am home')
```

```python
# Revisión
master_yoda('We are ready')
```

#### PROBLEMA 33: 

Dada una lista de enteros, retornar True si el arreglo tiene un  3  seguido por un  3 .

    has_33([1, 3, 3]) → True
    has_33([1, 3, 1, 3]) → False
    has_33([3, 1, 3]) → False

```python
def has_33(nums):
    cont = 0
    #Variable cont se inicializa en 0 y sirve como contador de cuantos 3 seguidos se han encontrado

    for n in nums:
        #Se analiza cada elemento de la lista
        #Si el numero es 3, se incrementa cont en 1
        if n == 3:
            cont += 1
        #Si el nUmero no es 3, se reinicia cont a 0 (porque se rompe la secuencia de dos numeros 3 seguidos.
        else:
            cont = 0
        if cont == 2:
            #En cuanto se detectan dos 3 seguidos, la funcion retorna True inmediatamente
            return True
    return False
```

```python
# Revisión
has_33([1, 3, 3])
```

```python
# Revisión
has_33([1, 3, 1, 3])
```

```python
# Revisión
has_33([3, 1, 3])
```

#### REPLICADOR: Dada una cadena, devuelve una cadena donde por cada carácter en el original hay tres caracteres REPETIDOS
    paper_doll('Hello') --> 'HHHeeellllllooo'
    paper_doll('Mississippi') --> 'MMMiiissssssiiippppppiii'

```python
def paper_doll(text):
    nuevapalabra = ""
    #La funcion recibe un texto
    for n in text:
        for i in range(1,4):
            #Por cada caracter en ese texto, se ejecuta un bucle interno que lo repite tres veces
            nuevapalabra += n
    
    return nuevapalabra
    
```

```python
# Check
paper_doll('Hello')
```

```python
# Check
paper_doll('Mississippi')
```

#### BLACKJACK: Dados tres enteros entre 1 y 11, si su suma es menor o igual a 21, devuelve su suma. Si su suma excede 21 *y* hay un once, reduce la suma total en 10. Finalmente, si la suma (incluso después del ajuste) excede 21, devuelve 'BUST'
    blackjack(5,6,7) --> 18
    blackjack(9,9,9) --> 'BUST'
    blackjack(9,9,11) --> 19

```python
def blackjack(a, b, c):
    total = a + b + c

    # Si hay un 11 y la suma pasa de 21, se reduce la suma total en 10
    if total > 21 and 11 in [a, b, c]:
        total -= 10

   #Si excede 21 ese total, devuelve "BUST"
    if total > 21:
        return 'BUST'
    else:
        return total
```

```python
# Revisión
blackjack(5,6,7)
```

```python
# Revisión
blackjack(9,9,9)
```

```python
# Revisión
blackjack(9,9,11)
```
