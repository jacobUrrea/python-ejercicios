Las condiciones hacen posible que un programa tome ciertas decisiones durante su ejecución.
En Python, las sentencias if, elif y else se usan para evaluar expresiones lógicas y realizar instrucciones únicamente cuando estas condiciones se cumplen.
Caracteristicas:
- Se evalúan como True o False
- Indentación obligatoria: Python usa sangría para definir bloques
- elif exclusivo: Solo se ejecuta un bloque que primero cumpla con la condicion
- else final: Captura todos los casos no cubiertos


Estructura basica:

```python
#if condicion:
    # codigo si la condicion es True
#elif otra condicion:  # opcional, multiples permitidos
    # codigo si otra condicion es True
#else:  # opcional
    # codigo si ninguna condicion fue True
```

Escribe una declaración if que determine si un disco musical tiene una calificación mayor a 8. Compruébalo utilizando el calificación del disco <b>“Back in Black”</b> el cual tiene un valor de 8.5. Si la condición es verdadera, imprime “"This album is Amazing!"”

```python
calificacion = 8.5

if calificacion > 8:
    print("This album is Amazing!")
    #Mediante esta condicion se evaluo si el valor de rating es mayor a 8
    #si esta condicion se cumple, se ejecutara el bloque del codigo que este dentro del if
    #Esto es bastante comun en situaciones en donde se deben filtrar elementos segun alguna caracteristica
    #por ejemplo, los sistemas de recomendacion
```

Escribe una declaración if-else que determine lo siguiente. Si la calificación es mayor que 8 imprime “this album is amazing”. Si es menor o igual a 8 que imprima “this album is ok”.

```python
if calificacion > 8:
    print("this album is amazing")
else:
    print("this album is ok")
#Si la calificacion es mayor a 8, este album sera excelente
#En otro caso, el album se calificara como aceptable
```

Escribe una declaración if que determine si un disco fue lanzado antes de 1980 o en los años 1991 o 1993. Si la condición es verdadera imprime el año de lanzamiento.

```python
añoLanzamiento = 1970
if añoLanzamiento < 1980 or (añoLanzamiento <= 1993 and añoLanzamiento >= 1993):
    print("El año de lanzamiento es:", añoLanzamiento)
```

```python
añoLanzamiento = 1992
if añoLanzamiento < 1980 or añoLanzamiento == 1993 or añoLanzamiento == 1991:
    print("El año de lanzamiento es:", añoLanzamiento)
```

Se utilizan operadores logicos como "or" para evaluar multiples condiciones dentro de una misma sentencia if. Si al menos una se cumple, se realizara la accion dentro del if.
Tambien esta la opcion de utilizar "and", para esto ambas condiciones deben cumplirse obligatoriamente, de lo contrario, la accion no se realizara.

```python
añoLanzamiento = 1992
if añoLanzamiento > 1980 and (añoLanzamiento <= 1993 or añoLanzamiento >= 1991):
    print("El año de lanzamiento es:", añoLanzamiento)

#Si el año de lanzamiento es mayor a 1980 Y esta entre 1993 y 1991, el sistema imprimira este año
```
