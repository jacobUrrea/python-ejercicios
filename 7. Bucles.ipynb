En este cuaderno se analiza el uso de bucles en Python, específicamente el uso de for y while.
Mediante los siguientes ejercicios se trabajará los recorridos de listas, el uso de rangos numéricos y el control de las condiciones para que un bucle finalice.
Destacando cómo los bucles permiten la automatización de procesos que requieren repetición de tareas.


Los bucles hacen posible repetir la ejecución de una tarea varias veces sin tener que escribir las mismas instrucciones una y otra vez.
El bucle for se emplea para iterar sobre secuencias, mientras que el uso de while continúa ejecutándose siempre que una condición específica se encuentre verdadera.

Caracteristicas:
- Iteración directa: Itera sobre elementos, no sobre índices, a diferencia de otros lenguajes.
- Iterables: Funciona con cualquier objeto iterable (listas, strings, rangos, diccionarios, etc.)
- Es mucho mas simple que los bucles for tradicionales de otros lenguajes

Casos de uso comunes:
- Procesar elementos de una colección
- Realizar operaciones repetitivas un número fijo de veces
- Transformar datos
- Buscar elementos en secuencias
- Recorrer estructuras de datos complejas

Escribe un bucle <code>for</code> que imprima todos los elementos entre <b>-5</b> y <b>5</b> usando la función range.

```python
for i in range(-5, 6):
    print(i)
#Esta funcion genera una secuencia de numeros desde -5 hasta 5. El valor final no se incluye
#Por eso es que se pone el 6 al final y no el 5
```

Imprime los elementos de la siguiente lista:
<code>Genres=[ 'rock', 'R&B', 'Soundtrack', 'R&B', 'soul', 'pop']</code>
Y asegúrate de seguir las convenciones del lenguaje Python

```python
# Lista de generos
Genres = ['rock', 'R&B', 'Soundtrack', 'R&B', 'soul', 'pop']
#Se recorre la lista usando un bucle for, es la forma mas sencilla 
for genre in Genres:
    print(genre)
```

Escribe un bucle for que imprima la siguiente lista: <code>squares=['red', 'yellow', 'green', 'purple', 'blue']</code>

```python
# Lista de colores
squares = ['red', 'yellow', 'green', 'purple', 'blue']

# Se imprime cada elemento 
for color in squares:
    print(color)
```

Escribe un bucle while que muestre los valores de la calificación de las canciones de un disco almacenados en la lista <code>PlayListRatings</code>. Si la calificación es menor a 6, sal del bucle. La lista <code>PlayListRatings</code> esta compuesta por: <code>PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]</code>

```python
# Lista de calificaciones
PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]
n=0
while PlayListRatings[n] >= 6:
    print(PlayListRatings[n])
    n += 1 
#Este bucle se detendra automaticamente cuando la calificaicon sea menor a 6
```

Escribe un bucle while que copie las cadenas <code>'orange'</code> de la lista <code>squares</code> a la lista <code>new_squares</code>. Detente y sal del ciclo si el valor en la lista no es <code>'orange'</code>:

```python
squares = ['orange', 'orange', 'purple', 'blue ', 'orange']
new_squares = []
n= 0
while squares[n] == "orange":
    new_squares.append(squares[n])
    n += 1

new_squares

#El bucle se ejecutara siempre que el elemento de la lista squares sea "orange"
#Cuando el valor sea diferente, el ciclo se detiene
```
