En este cuaderno se desarrollan ejercicios prácticos sobre el uso de tuplas en python, como lo es la indexacion, el slicing y la clasificacion de las tuplas

```python
# Creamos una tupla con diferentes generos musicales
genres_tuple = (
    "pop", "rock", "soul", "hard rock",
    "soft rock", "R&B", "progressive rock", "disco"
)

genres_tuple
#Las tuplas son inmutables (sus elementos no pueden modificarse una vez creados).
```

Encontrar la longitud de la tupla, <code>genres_tuple</code>:

```python
# len() devuelve la cantidad de elementos en la tupla
longitud = len(genres_tuple)
longitud
#El tamaño de la tupla sirve cuando se itera sobre esta misma sin conocer su tamaño exacto.
```

Accede al elemento en el índice 3:

```python
indice3 = genres_tuple[3]
indice3
# En Python, el primer elemento de una secuencia se accede con el indice 0, el segundo con el 1, el tercero con el 2, y asi sucesivamente
```

Utiliza el slicing para obtener los índices 3, 4 y 5:

```python
# El slicing extrae una sub-tupla sin modificar la original
indices = genres_tuple[3:6]
indices
#El indice final no se incluye
```

Encuentra los primeros dos elementos de la tupla <code>genres_tuple</code>:

```python
DosPrimeros = genres_tuple[:2]
DosPrimeros
#Cuando se trabaja con rankings, por ejemplo, esto resulta bastante comun 
```

Encuentra el primer índice de <code>"disco"</code>:

```python
disco = genres_tuple.index("disco")
disco
#index() devolvera la posicion de la primera aparicion del elemento
#Esto lanzara un error si el index no existe 
```

Genera una lista ordenada desde la tupla <code>C_tuple=(-5, 1, -3)</code>:

```python
# Se define la tupla
tupla = (-5, 1, -3)

# Se convierte a lista y se ordena
lista = sorted(tupla)
lista
#sorted() no modificara la lista original y siempre devolvera una lista, aun cuando se trabaje con tuplas
#Usar assert es una forma de comprobar si el resultado es correcto
#Ejemplo: assert C_list == [-5, -3, 1], "Error al ordenar la lista"
```
