Descripcion: 
En este cuaderno se aborda el uso de listas en Python, una de las estructuras de datos más flexibles que se pueden encontrar en cualquier lenguaje.
Se encontrarán ejercicios prácticos que incluyen creación de listas, el acceso mediante índices, slicing y concatenación, resaltando su valor para organizar y trabajar con diferentes cojuntos de datos.

Las listas en Python permiten almacenar varios elementos en una sola variable, al ccontrario de las tuplas,
las listas son estructuras capaces de alterarse después de su creación.
Esto las convierte en una herramienta fundamental para el desarrollo de algoritmos y aplicaciones reales.


Crea una lista <code>a_lit</code> con los siguientes elementos <code>1</code>, <code>hello</code>, <code>[1,2,3]</code> y <code>True</code>.

```python
lista = [1, "hello", [1, 2, 3], True]
lista
#Aca se demuestra como las listas pueden contener distintos tipos de datos, incluso otras listas 
#demostrando asi su flexibilidad
```

Encuentra el valor almacenado en el índice 1 de <code>a_list</code>.

```python
indice1 = lista[1]
indice1
#El indice 1 se refiere al segundo elemento, ya que como se menciono en los cuadernos anteriores, 
# Python comienza a contar desde el 0
```

Obtén el elemento almacenado en el índice 1, 2 y 3 de <code>a_list</code>.

```python
subLista = lista[1:4]
subLista
#No se incluye el ultimo indice especificado
```

Concatena las siguientes listas <code>A = [1, 'a']</code> y <code>B = [2, 1, 'd']</code>:

```python
# Definimos las listas
A = [1, 'a']
B = [2, 1, 'd']
# Concatenamos ambas listas
concatenada = A + B
concatenada
#Esta concatenacion crea una nueva lista, sin modificar las anteriores
```
