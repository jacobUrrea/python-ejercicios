En este cuaderno se analizara el uso de diccionarios, una estructura que organiza informacion mediante clave-valor. Se exploraran operaciones como la obtencion de claves y valores, el acceso directo a elementos y el uso de metodos para la gestion eficiente de datos estructurados.

Los diccionarios en python permite vincular una clave a un valor, esto permite acceder a esta informacion de manera rapida. Esto es la base del modelo de bases de datos sencillas, la definicion de configuraciones o el manejo de estructuras en las que cada elemento debe tener un identificador unico.

```python
soundtrack_dict = {"The Bodyguard":"1992", "Saturday Night Fever":"1977"}
soundtrack_dict 
```

a) En el diccionario <code>soundtrack_dict</code>. ¿Cuáles son las llaves?

```python
# Obtencion de las llaves del diccionario
llaves = soundtrack_dict.keys()
llaves
#Las llaves son la representacion de los identificadores unicos de cada elemento del diccionario
```

b) En el diccionario <code>soundtrack_dict</code>. ¿Cuáles son los valores?

```python
# Obtencion de los valores del diccionario
valores = soundtrack_dict.values()
valores
#Las llaves se pueden entender como los nombres de las columnas y los valores serian los datos asociados
```
